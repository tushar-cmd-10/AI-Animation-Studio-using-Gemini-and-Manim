```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel, Field
import os

os.environ["GOOGLE_API_KEY"] = "AQ.Ab8RN6JA2hVCZk7BwML8Jiicm3Tfchht6fJDn370NRepOEIuCA"

SYSTEM_PROMPT = """
You are an expert Manim (Community Edition) animation developer.

GOAL:
Convert the user's prompt into a COMPLETE, ERROR-FREE Manim Python script.

VISUAL & LAYOUT RULES (STRICT):
- Use ONLY relative positioning (.next_to, .arrange, .arrange_in_grid, .to_edge, .align_to).
- NEVER use absolute coordinates.
- NEVER allow overlaps.
- ALL content MUST stay inside the 16:9 frame.
- Respect config.frame_width and config.frame_height.
- Scale down objects or VGroups if needed.
- Maintain safe margins.
- Use readable font_size.
- Group elements using VGroup().

MANIM API SAFETY (MANDATORY):
- NEVER use .center, .top, .bottom attributes.
- ALWAYS use get_center(), get_top(), etc.
- NEVER pass method references to move_to(), shift(), align_to().
- NEVER call spatial methods on Scene (self.get_center() is INVALID).
- Scene has no geometry.
- Use ORIGIN, LEFT, RIGHT, UP, DOWN for frame alignment.
- Use .center() to center objects.

CODE RULES:
- Output ONLY valid Python code.
- Import exactly: from manim import *
- Scene class name MUST be GenScene.
- Use Create (not ShowCreation).
- Use Tex or MathTex only.
"""


class ManimCodeOutput(BaseModel):
    manim_code: str = Field(
        description="ONLY a complete, valid Manim Python script. No explanations."
    )

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", SYSTEM_PROMPT),
        ("human", "{input}")
    ]
)

llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0.0,
)

structured_llm = llm.with_structured_output(ManimCodeOutput)

chain = prompt | structured_llm

print("Manim GenAI Bot (Stateless) Ready! (Type 'exit' to quit)")

while True:
    user_prompt = input("\nUser: ")
    if user_prompt.lower() == "exit":
        break

    result = chain.invoke({"input": user_prompt})

    print("\nManim Code:\n")
    print(result.manim_code)
output_filename = "animation.py"

with open(output_filename, "w", encoding="utf-8") as f:
    f.write(result.manim_code)

print(f"\nManim code saved to: {output_filename}")
```
